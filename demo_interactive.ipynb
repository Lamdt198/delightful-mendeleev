## PHẦN 1: TIỀN XỬ LÝ & CHUẨN HÓA DỮ LIỆU VĂN BẢN (SLIDE 3.2 & 3.3)
### 1.1 Loại bỏ HTML tags và Boilerplate rác

```python
import re
from bs4 import BeautifulSoup

raw_html = """
<html>
    <body>
        <header><div>HEADER: Banner quang cao</div></header>
        <article>
            <h1>Hội nghị Khoa học Dữ liệu 2026</h1>
            <p>Hội thảo về <b>Trí tuệ nhân tạo</b> và <b>Kiến tạo tri thức</b>.</p>
        </article>
        <footer><p>© 2026 Viện Nghiên Cứu</p></footer>
    </body>
</html>
"""

soup = BeautifulSoup(raw_html, "html.parser")
for tag in soup(["header", "footer", "script", "style"]):
    tag.decompose()

clean_text = soup.get_text(separator=" ", strip=True)
print("Văn bản sau khi làm sạch:")
print(clean_text)
```

### 1.2 Tách câu dựa trên ngữ cảnh (Không làm gãy Tp. HCM, GS. TS)

```python
paragraph = "Ngày 15/09, GS. TS Hoàng Minh Sơn đã chủ trì hội thảo tại Tp. HCM. Sự kiện thu hút 500 sinh viên, v.v. Tất cả đều rất hào hứng."

abbreviations = [r"GS\.", r"TS\.", r"Tp\.", r"TP\.", r"v\.v\."]
protected = paragraph
for abbr in abbreviations:
    protected = re.sub(abbr, lambda m: m.group(0).replace(".", "<DOT>"), protected, flags=re.IGNORECASE)

sentences = [s.replace("<DOT>", ".").strip() for s in re.split(r'(?<=[.!?])\s+', protected) if s.strip()]

print(f"Tách thành {len(sentences)} câu chính xác:")
for i, s in enumerate(sentences, 1):
    print(f"  {i}. {s}")
```

### 1.3 Chuẩn hóa bộ mã tiếng Việt Unicode NFC vs NFD

```python
import unicodedata

s_nfc = unicodedata.normalize("NFC", "hòa")
s_nfd = unicodedata.normalize("NFD", "hòa")

print(f"s_nfc == s_nfd: {s_nfc == s_nfd} (Mắt thấy giống nhưng so sánh bằng máy báo SAI)")

# Chuẩn hóa về NFC:
fixed_nfd = unicodedata.normalize("NFC", s_nfd)
print(f"Sau khi chuẩn hóa NFC: s_nfc == fixed_nfd: {s_nfc == fixed_nfd}")
```
